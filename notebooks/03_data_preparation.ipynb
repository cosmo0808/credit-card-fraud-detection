# 3. Data Preparation

Use a fixed-seed, duplicate-aware stratified group split. Identical feature vectors stay in one partition; no resampling or feature selection is applied.

```python
from pathlib import Path
import sys
ROOT=Path.cwd()
if not (ROOT/"src").exists(): ROOT=ROOT.parent
sys.path.insert(0,str(ROOT))
from src.data_validation import load_transactions, duplicate_split_audit
from src.train import group_split
df=load_transactions(ROOT/"data"/"creditcard.csv")
X=df.drop(columns="Class"); y=df["Class"].astype(int)
X_train,X_test,y_train,y_test=group_split(X,y)
print(X_train.shape,X_test.shape); print("Fraud prevalence:",y_train.mean(),y_test.mean())
print("Split duplicate audit:",duplicate_split_audit(df,X_train.index,X_test.index))
```

Time and Amount are scaled only for Logistic Regression; PCA features are already transformed. Random Forest uses raw values. Scaling is inside a Pipeline and is fit on training rows only. Duplicates remain; the grouped split prevents identical feature rows from crossing partitions. Threshold comparisons use a separate grouped validation subset from training data. The final holdout is reserved for one final evaluation at the fixed 0.5 baseline threshold.
